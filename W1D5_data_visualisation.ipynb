```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

print("Libraries imported successfully")
```

```python
# Step 2: Create a sample dataset

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Hours_Studied": [2, 4, 5, 6, 7, 8, 9, 10],
    "Scores": [45, 55, 60, 65, 70, 78, 85, 92]
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
df
```

```python
# Step 3: Create a line plot
# Shows the relationship between study hours and scores

plt.figure(figsize=(8, 5))

plt.plot(df["Hours_Studied"], df["Scores"], marker="o")

plt.title("Study Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")

plt.grid(True)
plt.show()
```

```python
# Step 4: Create a bar chart
# Compares the scores of each student

plt.figure(figsize=(8, 5))

plt.bar(df["Student"], df["Scores"])

plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")

plt.show()
```

```python
# Step 5: Create a histogram
# Shows the distribution of student scores

plt.figure(figsize=(8, 5))

plt.hist(df["Scores"], bins=5, edgecolor="black")

plt.title("Distribution of Student Scores")
plt.xlabel("Scores")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Step 6: Create a scatter plot
# Shows the relationship between study hours and scores

plt.figure(figsize=(8, 5))

plt.scatter(df["Hours_Studied"], df["Scores"])

plt.title("Study Hours vs Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Scores")

plt.grid(True)
plt.show()
```

```python
# Step 7: Create a Seaborn distribution plot
# Visualises the distribution of student scores

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Scores", bins=5, kde=True)

plt.title("Distribution of Student Scores")
plt.xlabel("Scores")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Step 8: Create a box plot
# Shows the spread and possible outliers in student scores

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y="Scores")

plt.title("Distribution of Student Scores")
plt.ylabel("Scores")

plt.show()
```

```python
# Step 9: Create a correlation heatmap
# Shows the correlation between numerical variables

correlation = df[["Hours_Studied", "Scores"]].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Step 10: Display key insights from the visualisations

print("Key Insights:")
print("1. Study hours and scores show a strong positive relationship.")
print("2. Students who studied more hours generally achieved higher scores.")
print("3. The score distribution can be observed using the histogram.")
print("4. The box plot helps identify the spread of scores and possible outliers.")
```

## Final EDA Summary

- Study hours and scores show a strong positive relationship.
- Students who studied more hours generally achieved higher scores.
- Histogram shows the distribution of student scores.
- Box plot helps identify score spread and possible outliers.
