```python
import pandas as pd
import numpy as np

print("Libraries imported successfully")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Load the Indian dataset
df = pd.read_csv("w1d2_pandas/data/indian_population.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# Inspect missing values and duplicate rows
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Clean percentage values and convert them to numeric
df["growth_2001_2011"] = (
    df["growth_2001_2011"]
    .str.replace("%", "", regex=False)
    .astype(float)
)

print("Updated data types:")
print(df.dtypes)
```

```python
# Convert sex ratio from text to numeric
df["sex_ratio"] = pd.to_numeric(df["sex_ratio"], errors="coerce")

print("Updated data types:")
print(df.dtypes)
```

```python
# Verify the cleaned dataset
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:", df.duplicated().sum())

print("\nCleaned dataset:")
display(df.head(10))
```

```python
# Test: verify the dataset loaded correctly
assert df.shape == (36, 7)
assert df["state"].notna().all()
assert df["population"].notna().all()

print("All tests passed successfully.")
```

```python
# Export the cleaned dataset to CSV
df.to_csv("cleaned_indian_population.csv", index=False)

print("Cleaned dataset exported successfully.")
```

```python
# Test: verify the exported CSV
df_check = pd.read_csv("cleaned_indian_population.csv")

assert df_check.shape == df.shape
assert list(df_check.columns) == list(df.columns)

print("CSV export test passed successfully.")
```
