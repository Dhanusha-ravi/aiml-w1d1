```python
import pandas as pd

# Load CSV file and skip the description row
df = pd.read_csv("student_scores.csv", skiprows=1)

# Display dataset information
print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 rows:")
display(df.head(10))
```

```python
# Filter: select students who scored 70 or above
filtered_df = df[df["marks"] >= 70]

print("Students with marks >= 70:")
display(filtered_df)
```

```python
# Groupby: calculate average marks for each number of study hours
grouped_df = df.groupby("Hours_studied")["marks"].mean()

print("Average marks by study hours:")
display(grouped_df)
```

```python
# Merge: create a second DataFrame with student categories
student_info = pd.DataFrame({
    "Hours_studied": [2, 3, 4, 5, 6, 7, 8, 9, 10],
    "Category": ["Low", "Low", "Low", "Medium", "Medium",
                 "High", "High", "High", "High"]
})

merged_df = pd.merge(df, student_info, on="Hours_studied", how="inner")

print("Merged DataFrame:")
display(merged_df)
```

```python
# Pivot table: show average marks for each student category
pivot_df = pd.pivot_table(
    merged_df,
    values="marks",
    index="Category",
    aggfunc="mean"
)

print("Average marks by category:")
display(pivot_df)
```

```python
df.to_parquet("cleaned_student_scores.parquet", index=False)

print("Parquet file created successfully.")
```
