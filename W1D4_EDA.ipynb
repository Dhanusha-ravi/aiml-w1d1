```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("w1d2_pandas/data/indian_population.csv")

df["growth_2001_2011"] = df["growth_2001_2011"].str.replace("%", "").astype(float)

numeric_columns = [
    "population",
    "growth_2001_2011",
    "population_rural",
    "population_urban",
    "area"
]

df.head()
```

```python
correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45, ha="right")
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.colorbar(label="Correlation")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
top_10_states = df["state"].value_counts().head(10)

top_10_states
```

```python
top_10_states.plot(kind="bar", figsize=(10, 6))

plt.title("Top 10 States by Record Count")
plt.xlabel("State")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## EDA Narrative

The dataset contains population information for 36 Indian states and union territories. During the initial exploration, the dataset was checked using `describe()`, `info()`, and `isnull().sum()`. There are no missing values or duplicate records in the dataset. The `growth_2001_2011` column was originally stored as text with percentage symbols, so it was converted into a numeric column for analysis.

The population values vary considerably across states. Uttar Pradesh has the highest population in the dataset, while some states and union territories have much smaller populations. The distribution of population is therefore highly uneven and shows a strong skew toward a few highly populated states. Rural and urban population are strongly correlated with total population, with correlations of approximately 0.98 and 0.88 respectively. Area also has a positive relationship with population, while population growth has weak correlations with the other numerical variables.

The area column also varies widely between states, indicating possible outliers that should be considered during modelling. The `state` column contains unique state-level records, so the top-10 category count plot shows a count of one for each state and is not particularly informative for frequency analysis. Overall, the dataset is suitable for further analysis after checking distributions, scale differences, and potential outliers.
